```python
import pandas as pd
import numpy as np

# Load the datasets
ratings_df = pd.read_csv('top_500_ratings.csv')
anime_df = pd.read_csv('top_500_anime.csv')

print("Original ratings shape:", ratings_df.shape)

def reduce_ratings_dataset(ratings_df, min_user_ratings=10, min_anime_ratings=100, 
                         start_year=2013, sample_size=None):
    print("\nReduction Process:")
    print("-" * 50)
    
    # 1. Filter by recent years (2013 onwards for more recent, relevant data)
    print(f"\nStep 1: Filtering ratings from {start_year} onwards")
    ratings_df = ratings_df[ratings_df['watch_year'] >= start_year]
    print(f"Ratings after year filtering: {len(ratings_df):,}")
    
    # 2. Remove users with too few ratings
    user_rating_counts = ratings_df['username'].value_counts()
    active_users = user_rating_counts[user_rating_counts >= min_user_ratings].index
    ratings_df = ratings_df[ratings_df['username'].isin(active_users)]
    print(f"Ratings after filtering users with < {min_user_ratings} ratings: {len(ratings_df):,}")
    
    # 3. Remove anime with too few ratings
    anime_rating_counts = ratings_df['anime_id'].value_counts()
    popular_anime = anime_rating_counts[anime_rating_counts >= min_anime_ratings].index
    ratings_df = ratings_df[ratings_df['anime_id'].isin(popular_anime)]
    print(f"Ratings after filtering anime with < {min_anime_ratings} ratings: {len(ratings_df):,}")
    
    # 4. Stratified sampling to maintain rating distribution
    if sample_size:
        print(f"\nPerforming stratified sampling to get {sample_size:,} ratings")
        # Calculate the proportion of each rating
        rating_proportions = ratings_df['rating'].value_counts(normalize=True)
        
        sampled_ratings = []
        for rating, proportion in rating_proportions.items():
            rating_group = ratings_df[ratings_df['rating'] == rating]
            n_samples = int(sample_size * proportion)
            sampled = rating_group.sample(n=min(n_samples, len(rating_group)))
            sampled_ratings.append(sampled)
        
        ratings_df = pd.concat(sampled_ratings)
        print(f"Final number of ratings after sampling: {len(ratings_df):,}")
    
    # Print final statistics
    print("\nFinal Dataset Statistics:")
    print("-" * 50)
    print(f"Number of unique users: {ratings_df['username'].nunique():,}")
    print(f"Number of unique anime: {ratings_df['anime_id'].nunique():,}")
    print("\nRating Distribution:")
    rating_dist = ratings_df['rating'].value_counts().sort_index()
    for rating, count in rating_dist.items():
        print(f"Rating {rating}: {count:,} ({count/len(ratings_df)*100:.2f}%)")
    
    return ratings_df

# Reduce the dataset with specific parameters
reduced_ratings = reduce_ratings_dataset(
    ratings_df,
    min_user_ratings=20,      # Users must have rated at least 20 anime
    min_anime_ratings=200,    # Anime must have at least 200 ratings
    start_year=2013,         # Only include ratings from 2013 onwards
    sample_size=5_000_000    # Target approximately 5 million final ratings
)

# Save the reduced dataset
reduced_ratings.to_csv('reduced_top_500_ratings.csv', index=False)
print("\nSaved reduced dataset to 'reduced_top_500_ratings.csv'")

# Print year distribution of the reduced dataset
print("\nYear Distribution in Reduced Dataset:")
print("-" * 50)
year_dist = reduced_ratings['watch_year'].value_counts().sort_index()
for year, count in year_dist.items():
    print(f"Year {year}: {count:,} ratings ({count/len(reduced_ratings)*100:.2f}%)")
```

```python
import pandas as pd
import numpy as np
from datetime import datetime

# Load the reduced ratings dataset
ratings_df = pd.read_csv('top_500_ratings.csv')

# Convert date columns to datetime
ratings_df['my_last_updated'] = pd.to_datetime(ratings_df['my_last_updated'])
ratings_df['my_start_date'] = pd.to_datetime(ratings_df['my_start_date'])
ratings_df['my_finish_date'] = pd.to_datetime(ratings_df['my_finish_date'])

# Calculate global mean rating
global_mean = ratings_df['rating'].mean()

# Calculate user and anime biases
user_mean = ratings_df.groupby('username')['rating'].mean()
user_bias = user_mean - global_mean
anime_mean = ratings_df.groupby('anime_id')['rating'].mean()
anime_bias = anime_mean - global_mean

# Add bias-adjusted rating
ratings_df['user_bias'] = ratings_df['username'].map(user_bias)
ratings_df['anime_bias'] = ratings_df['anime_id'].map(anime_bias)
ratings_df['adjusted_rating'] = ratings_df['rating'] - ratings_df['user_bias'] - ratings_df['anime_bias']

# Add user engagement features
user_stats = ratings_df.groupby('username').agg({
    'anime_id': 'count',
    'rating': ['mean', 'std'],
    'my_rewatching': 'sum'
}).reset_index()
user_stats.columns = ['username', 'total_ratings', 'mean_rating', 'rating_std', 'total_rewatches']

# Merge user stats back to ratings
ratings_df = ratings_df.merge(user_stats, on='username', how='left')

# Save preprocessed ratings
ratings_df.to_csv('preprocessed_reduced_ratings.csv', index=False)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load the anime dataset
anime_df = pd.read_csv('top_500_anime.csv')

# Convert aired_from_year to int and handle any missing values
anime_df['aired_from_year'] = pd.to_numeric(anime_df['aired_from_year'], errors='coerce')
anime_df['aired_from_year'] = anime_df['aired_from_year'].fillna(anime_df['aired_from_year'].median())

# Create popularity score combining members, favorites, and score
scaler = MinMaxScaler()
anime_df['popularity_score'] = scaler.fit_transform(
    anime_df[['members', 'favorites', 'score']].values
).mean(axis=1)

# Convert type to dummy variables
type_dummies = pd.get_dummies(anime_df['type'], prefix='type')
anime_df = pd.concat([anime_df, type_dummies], axis=1)

# Convert source to dummy variables
source_dummies = pd.get_dummies(anime_df['source'], prefix='source')
anime_df = pd.concat([anime_df, source_dummies], axis=1)

# Normalize numerical features
numerical_cols = ['episodes', 'duration_min', 'score', 'scored_by', 
                 'members', 'favorites', 'popularity']
anime_df[numerical_cols] = scaler.fit_transform(anime_df[numerical_cols])

# Save preprocessed anime data
anime_df.to_csv('preprocessed_reduced_anime.csv', index=False)
```

```python
import pandas as pd
from datetime import datetime

# Load the users dataset
users_df = pd.read_csv('top_500_users.csv')

# Convert date columns to datetime
date_columns = ['birth_date', 'join_date', 'last_online']
for col in date_columns:
    users_df[col] = pd.to_datetime(users_df[col])

# Calculate account age in days
users_df['account_age_days'] = (datetime.now() - users_df['join_date']).dt.days

# Calculate activity ratio
users_df['activity_ratio'] = (users_df['user_completed'] + users_df['user_watching']) / \
                            (users_df['user_completed'] + users_df['user_watching'] + 
                             users_df['user_dropped'] + users_df['user_plantowatch'])

# Create engagement score
users_df['engagement_score'] = (
    users_df['user_days_spent_watching'] * 0.3 +
    users_df['stats_episodes'] * 0.3 +
    users_df['user_completed'] * 0.2 +
    users_df['stats_rewatched'] * 0.2
)

# Convert gender to dummy variables
gender_dummies = pd.get_dummies(users_df['gender'], prefix='gender')
users_df = pd.concat([users_df, gender_dummies], axis=1)

# Normalize numerical features
numerical_cols = ['user_days_spent_watching', 'stats_episodes', 'stats_mean_score',
                 'user_completed', 'user_watching', 'user_dropped', 'engagement_score']
scaler = MinMaxScaler()
users_df[numerical_cols] = scaler.fit_transform(users_df[numerical_cols])

# Save preprocessed users data
users_df.to_csv('preprocessed_reduced_users.csv', index=False)
```

```python
import pandas as pd
import numpy as np

# Load the datasets
ratings_df = pd.read_csv('preprocessed_reduced_ratings.csv')
anime_df = pd.read_csv('preprocessed_reduced_anime.csv')
users_df = pd.read_csv('preprocessed_reduced_users.csv')

# Function to print detailed analysis
def analyze_dataset(df, name):
    print(f"\n{'='*50}")
    print(f"{name} Dataset Analysis")
    print(f"{'='*50}")
    
    # Basic information
    print("\nDataset Info:")
    print(f"Number of rows: {len(df)}")
    print(f"Number of columns: {len(df.columns)}")
    print("\nColumns:", df.columns.tolist())
    
    # Missing values
    print("\nMissing Values:")
    print(df.isnull().sum())
    
    # Numerical columns description
    print("\nNumerical Columns Statistics:")
    print(df.describe())
    
    # Data types
    print("\nData Types:")
    print(df.dtypes)
    
    if 'rating' in df.columns:
        print("\nRating Distribution:")
        print(df['rating'].value_counts().sort_index())

# Analyze each dataset
analyze_dataset(ratings_df, "Ratings")
analyze_dataset(anime_df, "Anime")
analyze_dataset(users_df, "Users")

# Additional analysis for ratings
print("\nRatings per User Statistics:")
ratings_per_user = ratings_df.groupby('username').size()
print(ratings_per_user.describe())

print("\nRatings per Anime Statistics:")
ratings_per_anime = ratings_df.groupby('anime_id').size()
print(ratings_per_anime.describe())

# Check for any potential data quality issues
def check_data_quality(ratings_df, anime_df, users_df):
    print("\nData Quality Checks:")
    
    # Check rating range
    print(f"\nRating range: {ratings_df['rating'].min()} to {ratings_df['rating'].max()}")
    
    # Check for orphaned ratings (ratings for non-existent anime or users)
    missing_anime = ratings_df[~ratings_df['anime_id'].isin(anime_df['anime_id'])]
    missing_users = ratings_df[~ratings_df['username'].isin(users_df['username'])]
    
    print(f"\nOrphaned ratings (missing anime): {len(missing_anime)}")
    print(f"Orphaned ratings (missing users): {len(missing_users)}")
    
    # Check for duplicate ratings
    duplicates = ratings_df.groupby(['username', 'anime_id']).size()
    print(f"\nUsers with multiple ratings for same anime: {len(duplicates[duplicates > 1])}")

check_data_quality(ratings_df, anime_df, users_df)
```

```python
# Drop columns with too many missing values
ratings_df = ratings_df.drop(['my_tags'], axis=1)

# Fill missing rewatching values
ratings_df['my_rewatching'] = ratings_df['my_rewatching'].fillna(0)

# Fill missing rating_std
ratings_df['rating_std'] = ratings_df['rating_std'].fillna(ratings_df['rating_std'].mean())
```

```python
# Add a normalized rating column
def normalize_ratings(df):
    df['normalized_rating'] = (df['rating'] - df['rating'].mean()) / df['rating'].std()
    return df

ratings_df = normalize_ratings(ratings_df)
```

```python
# Remove orphaned ratings
valid_users = users_df['username'].unique()
ratings_df = ratings_df[ratings_df['username'].isin(valid_users)]

# Fill missing activity_ratio
users_df['activity_ratio'] = users_df['activity_ratio'].fillna(users_df['activity_ratio'].mean())
```

```python
# Add interaction frequency features
ratings_df['user_rating_count'] = ratings_df.groupby('username')['rating'].transform('count')
ratings_df['anime_rating_count'] = ratings_df.groupby('anime_id')['rating'].transform('count')

# Add temporal features
ratings_df['days_since_start'] = (pd.to_datetime(ratings_df['my_finish_date']) - 
                                 pd.to_datetime(ratings_df['my_start_date'])).dt.days
```

```python
# Remove constant columns from anime dataset
anime_cols_to_drop = ['Yaoi', 'Yuri', 'Unknown']
anime_df = anime_df.drop(anime_cols_to_drop, axis=1)
```

```python
# Cap extreme values in user statistics
def cap_outliers(df, column, lower_percentile=1, upper_percentile=99):
    lower = np.percentile(df[column], lower_percentile)
    upper = np.percentile(df[column], upper_percentile)
    df[column] = df[column].clip(lower, upper)
    return df

columns_to_cap = ['user_days_spent_watching', 'stats_rewatched', 'stats_episodes']
for col in columns_to_cap:
    users_df = cap_outliers(users_df, col)
```

```python
# Save the final preprocessed datasets
try:
    # Save ratings dataset
    ratings_df.to_csv('final_ratings.csv', index=False)
    print("✅ Successfully saved final_ratings.csv")
    
    # Save anime dataset
    anime_df.to_csv('final_anime.csv', index=False)
    print("✅ Successfully saved final_anime.csv")
    
    # Save users dataset
    users_df.to_csv('final_users.csv', index=False)
    print("✅ Successfully saved final_users.csv")
    
    # Print dataset sizes for verification
    print("\nDataset Sizes:")
    print(f"Ratings: {len(ratings_df):,} rows")
    print(f"Anime: {len(anime_df):,} rows")
    print(f"Users: {len(users_df):,} rows")
    
except Exception as e:
    print(f"❌ Error saving files: {str(e)}")
    
```
